#  **tokenize the en_tokens**

```python
def tokenize_example(example, en_nlp, fr_nlp, max_length, lower, sos_token, eos_token):
    en_tokens = [token.text for token in en_nlp.tokenizer(example["translation"]["en"])][:max_length]
    fr_tokens = [token.text for token in fr_nlp.tokenizer(example["translation"]["fr"])][:max_length]
    if lower:
        en_tokens = [token.lower() for token in en_tokens]
        fr_tokens = [token.lower() for token in fr_tokens]
    en_tokens = [sos_token] + en_tokens + [eos_token]
    fr_tokens = [sos_token] + fr_tokens + [eos_token]
    return {"en_tokens": en_tokens, "fr_tokens": fr_tokens}
```

```python
def get_collate_fn(pad_index):
    def collate_fn(batch):
        batch_en_ids = [example["en_ids"] for example in batch]
        batch_fr_ids = [example["fr_ids"] for example in batch]
        batch_en_ids = nn.utils.rnn.pad_sequence(batch_en_ids, padding_value=pad_index)
        batch_fr_ids = nn.utils.rnn.pad_sequence(batch_fr_ids, padding_value=pad_index)
        batch = {
            "en_ids": batch_en_ids,
            "fr_ids": batch_fr_ids,
        }
        return batch

    return collate_fn
```

```python
def get_data_loader(dataset, batch_size, pad_index, shuffle=False):
    collate_fn = get_collate_fn(pad_index)
    data_loader = torch.utils.data.DataLoader(
        dataset=dataset,
        batch_size=batch_size,
        collate_fn=collate_fn,
        shuffle=shuffle,
    )
    return data_loader
```

# The model

*   encoder

**encoder with one Lstm so he will learn and send the hidden and the cell for the different LSTM of the decoder
we use input_dim and embedding_dim for the self.embedding  Converts input indices (word tokens) into dense vectors of size embedding_dim.
we did nn.LSTM(embedding_dim, hidden_dim, n_layers, dropout=dropout) Processes the embedded input sequence and captures long-term dependencies in the data.
we add dropout to make the model preform better and genralize better(helps prevent overfitting by randomly deactivating neurons during training.)**

```python
class Encoder(nn.Module):
    def __init__(self, input_dim, embedding_dim, hidden_dim, n_layers, dropout):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.n_layers = n_layers
        #self.embedding = nn.Embedding(input_dim, embedding_dim, padding_idx=pad_index)
        self.embedding = nn.Embedding(input_dim, embedding_dim)
        self.rnn  = nn.LSTM(embedding_dim, hidden_dim, n_layers, dropout=dropout)
        self.dropout = nn.Dropout(dropout)

    def forward(self, src):
        # src = [src length, batch size]
        embedded = self.dropout(self.embedding(src))
        
        # embedded = [src length, batch size, embedding dim]
        outputs, (hidden, cell) = self.rnn(embedded)
        
        # outputs = [src length, batch size, hidden dim * n directions]
        # hidden = [n layers * n directions, batch size, hidden dim]
        # cell = [n layers * n directions, batch size, hidden dim]
        # outputs are always from the top hidden layer
        return hidden, cell
```

* Decoder

**different LSTM 
  the embedding Layers- Converts target sequence tokens into dense embedding vectors.
  nn.LSTM(embedding_dim, hidden_dim, n_layers, dropout=dropout)- Processes the embedded input one token at a time (during decoding).
  self.fc_out = nn.Linear(hidden_dim, output_dim)- maps the hidden state output to the target vocabulary size to predict the next token.
  dropout-helps prevent overfitting by randomly deactivating neurons during training.
  and it send the prediction he did for what hidden state and cell state**

```python
class Decoder(nn.Module):
    def __init__(self, output_dim, embedding_dim, hidden_dim, n_layers, dropout):
        super().__init__()
        self.output_dim = output_dim
        self.hidden_dim = hidden_dim
        self.n_layers = n_layers
        #self.embedding = nn.Embedding(output_dim, embedding_dim, padding_idx=pad_index)
        self.embedding = nn.Embedding(output_dim, embedding_dim)
        self.rnn = nn.LSTM(embedding_dim, hidden_dim, n_layers, dropout=dropout)
        self.fc_out = nn.Linear(hidden_dim, output_dim)
        self.dropout = nn.Dropout(dropout)

    def forward(self, input, hidden, cell):
        # input = [batch size]
        # hidden = [n layers * n directions, batch size, hidden dim]
        # cell = [n layers * n directions, batch size, hidden dim]
        # n directions in the decoder will both always be 1, therefore:
        # hidden = [n layers, batch size, hidden dim]
        # context = [n layers, batch size, hidden dim]
        input = input.unsqueeze(0)
        
        # input = [1, batch size]
        embedded = self.dropout(self.embedding(input))
        
        # embedded = [1, batch size, embedding dim]
        output, (hidden, cell) = self.rnn(embedded, (hidden, cell))
        
        # output = [seq length, batch size, hidden dim * n directions]
        # hidden = [n layers * n directions, batch size, hidden dim]
        # cell = [n layers * n directions, batch size, hidden dim]
        # seq length and n directions will always be 1 in this decoder, therefore:
        # output = [1, batch size, hidden dim]
        # hidden = [n layers, batch size, hidden dim]
        # cell = [n layers, batch size, hidden dim]
        
        prediction = self.fc_out(output.squeeze(0))

        return prediction, hidden, cell
```

* Seq2Seq

**we in the model use the encoder decoder to predict the sequence we have mode for train with greedy method- teacher forcing that try to find the greedy predict to the sequence in the train
and in the infer mode we will try to decode with the beam_search_decode to find the best predict for the sequence in infer time after we done train the model(more explain about this in the the cell after the model)**

```python
import torch
import torch.nn as nn

class Seq2Seq(nn.Module):
    def __init__(self, encoder, decoder, device, beam_width):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder
        self.device = device
        self.beam_width = beam_width
        assert encoder.hidden_dim == decoder.hidden_dim, "Hidden dimensions must be equal!"
        assert encoder.n_layers == decoder.n_layers, "Encoder and decoder must have equal number of layers!"
    def forward(self, src, trg=None, teacher_forcing_ratio=0.5, mode="train"):
        """
        mode="train" -> Uses teacher forcing
        mode="infer" -> Uses beam search for decoding
        """
        if mode == "train":
            return self.train_forward(src, trg, teacher_forcing_ratio)
        elif mode == "infer":
            return self.beam_search_decode(src)

    def train_forward(self, src, trg, teacher_forcing_ratio=0.5):#we use greedy method as 
        batch_size = trg.shape[1]
        trg_length = trg.shape[0]
        trg_vocab_size = self.decoder.output_dim
    
        outputs = torch.zeros(trg_length, batch_size, trg_vocab_size).to(self.device)
    
        hidden, cell = self.encoder(src)
        input = trg[0, :]  # First input is <sos>
    
        for t in range(1, trg_length):
            output, hidden, cell = self.decoder(input, hidden, cell)
            outputs[t] = output
    
            # Use teacher forcing
            teacher_force = torch.rand(1).item() < teacher_forcing_ratio
            top1 = output.argmax(1)  # Get the most likely token
    
            input = trg[t] if teacher_force else top1  # Use true word or predicted word
    
        return outputs

    #def beam_search_decode(self, src):
    #    trg_vocab_size = self.decoder.output_dim
    #    eos_token = fr_vocab['<eos>']  # EOS token ID

     #    max_len = 200  # Prevent infinite loops
    
     #    hidden, cell = self.encoder(src)
     #    input = torch.tensor([2]).to(self.device)  # Start with <sos> token
    
     #    beams = [(input, hidden, cell, 0, [2])]  # (token, hidden, cell, score, sequence)
     #    final_sequences = []
    
     #    for t in range(1, max_len):
     #        all_candidates = []
    
     #        for seq_input, seq_hidden, seq_cell, seq_score, seq_sequence in beams:
     #            if seq_sequence[-1] == eos_token:
     #                final_sequences.append((seq_sequence, seq_score))
     #                continue
      #           # Decode step    
      #           output, hidden, cell = self.decoder(seq_input, seq_hidden, seq_cell)
      #           topk_probs, topk_tokens = torch.topk(output, self.beam_width, dim=1)
    
      #           for i in range(self.beam_width):
       #              token = topk_tokens[:, i]
       #              token_prob = topk_probs[:, i]
       #              new_score = seq_score + token_prob[0].item()
       #             new_sequence = seq_sequence + [token[0].item()]
       #              if token[0].item() == fr_vocab[eos_token]:  # <eos> token
       #                  final_sequences.append((new_sequence, new_score))
       #                  continue
        #             else:
       #                  all_candidates.append((token, hidden, cell, new_score, new_sequence))
    # 
            # Stop if all beams predict <eos>
    #         if not all_candidates:
    #             break
    
     #        all_candidates = sorted(all_candidates, key=lambda x: x[3], reverse=True)
     #        beams = all_candidates[:self.beam_width]  # Keep top beams
    # 
        # Get the best sequence
     #    if final_sequences:
    #         best_sequence = max(final_sequences, key=lambda x: x[1])[0]
     #    else:
     #        best_sequence = beams[0][4]
    
    #     return best_sequence
    def beam_search_decode(self, src ):
        """
        Perform beam search decoding for sequence generation.
        
        Args:
            src (torch.Tensor): Source sentence tensor (input sequence).
            beam_width (int): Number of beams to maintain.
            max_len (int): Maximum decoding length to prevent infinite loops.
        
        Returns:
            List[int]: The best generated sequence (token indices).
        """
        max_len=100
        trg_vocab_size = self.decoder.output_dim  # Target vocabulary size
        eos_token = fr_vocab['<eos>']  # EOS token ID
        sos_token = fr_vocab['<sos>']  # SOS token ID
    
        # Encode the source sequence
        hidden, cell = self.encoder(src)
    
        # Start decoding with <sos> token
        input = torch.tensor([sos_token]).to(self.device)
    
        # Beam structure: (token, hidden, cell, score, sequence)
        beams = [(input, hidden, cell, 0, [sos_token])]
        final_sequences = []
    
        for t in range(1, max_len):
            all_candidates = []
    
            for seq_input, seq_hidden, seq_cell, seq_score, seq_sequence in beams:
                # Stop expanding if the last token is <eos>
                if seq_sequence[-1] == eos_token:
                    final_sequences.append((seq_sequence, seq_score))
                    continue
    
                # Decode step
                output, hidden, cell = self.decoder(seq_input, seq_hidden, seq_cell)
                output = torch.log_softmax(output, dim=1)  # Use log probabilities for stability
                topk_probs, topk_tokens = torch.topk(output, self.beam_width, dim=1)
    
                for i in range(self.beam_width):
                    token = topk_tokens[:, i]
                    token_prob = topk_probs[:, i]
                    new_score = seq_score + token_prob[0].item()
                    new_sequence = seq_sequence + [token[0].item()]
    
                    all_candidates.append((token, hidden, cell, new_score, new_sequence))
    
            # Stop if no new candidates
            if not all_candidates:
                break
    
            # Keep top-k beams
            all_candidates = sorted(all_candidates, key=lambda x: x[3], reverse=True)
            beams = all_candidates[:self.beam_width]
    
            # Stop expanding if all sequences in beams end with <eos>
            if all(seq[-1] == eos_token for _, _, _, _, seq in beams):
                break
    
        # Get the best sequence based on the highest score
        if final_sequences:
            best_sequence = max(final_sequences, key=lambda x: x[1])[0]
        else:
            best_sequence = beams[0][4]  # If no sequence reaches <eos>, take the best beam
    
        return best_sequence

  
```

# The model parameters

**we chose them by learn different sizes for embedding layers and dropout to preform the best we saw in the tries we did**


```python
input_dim = len(en_vocab)#(15792 )
output_dim = len(fr_vocab)# 21335
encoder_embedding_dim = 1024  # Increased from 256  512
decoder_embedding_dim = 1024  # Increased from 256  512
hidden_dim = 1024  # Increased from 512  1024 ->370 val loss
n_layers = 4 #4 
# LAST EMBED=512 HIDDEN=1024 LAYERS=4
encoder_dropout = 0.3 # Reduced from 0.5
decoder_dropout = 0.3  # Reduced from 0.5
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

encoder = Encoder(
    input_dim,
    encoder_embedding_dim,
    hidden_dim,
    n_layers,
    encoder_dropout,
)

decoder = Decoder(
    output_dim,
    decoder_embedding_dim,
    hidden_dim,
    n_layers,
    decoder_dropout,
)

model = Seq2Seq(encoder, decoder, device, beam_width=2).to(device)

if torch.cuda.device_count() > 1:
    model = nn.DataParallel(model)
```

**criterion = nn.CrossEntropyLoss(ignore_index=pad_index) It is commonly used for multi-class classification tasks, such as selecting the correct word from a vocabulary during sequence generation.
pad_index is the index of the <pad> token in your vocabulary.
In Seq2Seq models, sentences vary in length, so shorter sequences are padded with <pad> tokens.
However, padding tokens should not contribute to the loss, as they are meaningless.
ignore_index=pad_index tells CrossEntropyLoss to ignore the loss for padded tokens, preventing them from affecting gradient updates.**

```python
criterion = nn.CrossEntropyLoss(ignore_index=pad_index)
```

**train function on the tensor ids from english to french**

```python
import tqdm

def train_fn(model, batches, optimizer, criterion, clip, teacher_forcing_ratio, device):
    model.train()
    epoch_loss = 0

    # Create a progress bar for batches
    with tqdm.tqdm(total=len(batches), desc="Training Batches") as pbar:
        for batch in batches:  # Loop through the list of batches
            src = batch["en_ids"].to(device)
            trg = batch["fr_ids"].to(device)
            # src = [src length, batch size]
            # trg = [trg length, batch size]
            optimizer.zero_grad()
            output = model(src, trg, teacher_forcing_ratio, mode="train")
            # output = [trg length, batch size, trg vocab size]
            output_dim = output.shape[-1]
            output = output[1:].view(-1, output_dim)
            # output = [(trg length - 1) * batch size, trg vocab size]
            trg = trg[1:].view(-1)
            # trg = [(trg length - 1) * batch size]
            loss = criterion(output, trg)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), clip)
            optimizer.step()
            epoch_loss += loss.item()
            pbar.update(1)  # Update the progress bar after each batch

    return epoch_loss / len(batches)  # Average loss over batches
```

**evaluate_fn**

```python
def evaluate_fn(model, data_loader, criterion, device):
    model.eval()
    epoch_loss = 0
    with torch.no_grad():
        for i, batch in enumerate(data_loader):
            src = batch["en_ids"].to(device)
            trg = batch["fr_ids"].to(device)
            # src = [src length, batch size]
            # trg = [trg length, batch size]
            output = model(src, trg, 0)  # turn off teacher forcing
            # output = [trg length, batch size, trg vocab size]
            output_dim = output.shape[-1]
            output = output[1:].view(-1, output_dim)
            # output = [(trg length - 1) * batch size, trg vocab size]
            trg = trg[1:].view(-1)
            # trg = [(trg length - 1) * batch size]
            loss = criterion(output, trg)
            epoch_loss += loss.item()
    return epoch_loss / len(data_loader)
```

```python
def translate_sentence(
    sentence,
    model,
    en_nlp, fr_nlp,  # Tokenizers for English & French
    en_vocab, fr_vocab,  # Vocabulary mappings
    lower, sos_token, eos_token,
    device, max_output_length=100
):
    model.eval()  # Set model to evaluation mode

    with torch.no_grad():  # Disable gradient calculations
        # Tokenize the sentence
        if isinstance(sentence, str):
            tokens = [token.text for token in en_nlp.tokenizer(sentence)]
        else:
            tokens = [token for token in sentence]

        if lower:
            tokens = [token.lower() for token in tokens]

        # Add <sos> and <eos> tokens
        tokens = [sos_token] + tokens + [eos_token]
        print(f"Tokenized input: {tokens}")

        # Convert tokens to vocabulary indices
        ids = en_vocab.lookup_indices(tokens)
        tensor = torch.LongTensor(ids).unsqueeze(-1).to(device)  # Shape: [seq_len, 1]

        # ✅ Fix: Access the real model inside DataParallel
        best_sequence = model.module.beam_search_decode(tensor) if isinstance(model, torch.nn.DataParallel) else model.beam_search_decode(tensor)

        # Convert output indices to words
        translated_tokens = fr_vocab.lookup_tokens(best_sequence)
        print(f"Translated output: {translated_tokens}")

    return translated_tokens  # Return translated sentence as tokens
```

**get_tokenizer_fn so the tokens will be the same in lower and tokenizied**

```python
def get_tokenizer_fn(nlp, lower):
    def tokenizer_fn(s):
        
        tokens = [token.text for token in nlp.tokenizer(s)]
        if lower:
            tokens = [token.lower() for token in tokens]
        return tokens

    return tokenizer_fn
```
